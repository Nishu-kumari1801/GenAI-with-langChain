## magic methods

magic methods in python , also knoown as dunder method(double underscore method), are special methods that start and end with double underscores . These methods enable you to define the behaviour of objects for built-in operations , such as airthematic operations, comparisons and more.


## magic method
magic method are predefined methods in python that you can override to change the behaviour of your objects . some common magic methods include:
. init : initializes a new instance of a class.
. str : returns a string repersentation of an object.
. repr : returns an official string repersentation of an object
. len : returns the length oof an object 
. getitem : Gets an item from a  container
. setitem : sets an item in a container

```python
## basic methods
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age

person = Person("Nishu",21)
print(person)
```

```python
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    def __str__(self):
        return f"{self.name},{self.age} years old"

    def __repr__(self):
        return f"Person(name = {self.name},age={self.age})"
    
person = Person("Nishu",21)
print(person)
print(repr(person))
```
