### introduction to dictionaries

Dictionaries are unordered collections of items . They store data in key-value pairs . Keys must be unique and immutable (e.g., strings,numbers, ot tuples) , while values can be of any type.

```python
## creating Dictionaries 
empty_dict = {}
print(type(empty_dict))
```

```python
empty_dict=dict()
empty_dict
```

```python
student = {"name":"Nishu","age":21,"grade":"A+"}
print(student)
print(type(student))
```

```python
## Single key is always used
student = {"name":"Nishu","age":21,"name":"A+"}
print(student)
```

```python
###  accessing dictionary elements
student = {"name":"Nishu","age":21,"grade":"A+"}
print(student)
```

```python
print(student['grade'])
```

```python
print(student['age'])
```

```python
## accessing using get() method
print(student.get('grade'))
print(student.get('last_name'))
print(student.get('last_name',"Not Available"))
```

```python
##Modifying dictionary elements 
## dictionary are mutable, so you can add , update or delete elements
print(student)
```

```python
student["age"]=33   ## update the value for a key
print(student)
student["address"]="India" ## added a new key-value
print(student)
```

```python
del student['grade'] ## delete the key and value there
print(student)
```

```python
## Dictionary methods
keys = student.keys() ## get all the keys
print(keys)
values = student.values() ## get all the values
print(values)

items =  student.items() ## get all key value pairs (in form of list of tuples)
print(items)
```

```python
## shallow copy
student_copy = student
print(student)
print(student_copy)
```

```python
student['name']="jha"
print(student)
print(student_copy)
```

```python
student_copy1 = student.copy() ## shallow copy
print(student)
print(student_copy1)
```

```python
student["name"]="krish3"
print(student)
print(student_copy1)
```

```python
## Iterating over dictionaries
## we can use loops to iterate over dictionaries , keys, values , or items

## Iterating over keys 
for keys in student.keys():
    print(keys)

```

```python
## Iterating over values 
for keys in student.values():
    print(keys)
```

```python
## Iterate over key value pairs

for key,value in student.items():
    print(f"{key}:{value}")
```

```python
## Nested dictionaries 
students = {
    "student1" :{"name":"nishu","age":33},
    "student2" :{"name":"nitish","age":24}
}
print(students)
```

```python
## Access nested dictionaries
print(students["student2"]["name"])
print(students["student2"]["age"])
```

```python
students.items()
```

```python
## Iteratinng over nested dictionaries
for student_id,student_info in students.items():
    print(f"{student_id}:{student_info}")
    for key,value in student_info.items():
        print(f"{key}:{value}")
```

```python
## Dictionary comprehension

squares = {x:x**2 for x in range(5)}
print(squares)
```

```python
## conditional dictionaries comprehension 
evens = {x:x**2 for x in range(10) if x%2==0}
print(evens)
```

```python
## Practical examples 
## use a dictionary to count the frequency of element in list
numbers = [1,2,2,3,3,3,4,4,4,4]
frequency={}
for number in numbers:
    if number  in frequency:
        frequency[number]+=1
    else :
        frequency[number]=1
print(frequency)
```

```python
## merge 2 dictionaries into one

dict1 = {"a":1,"b":2}
dict2={"b":3,"c":4}
merged_dict = {**dict1,**dict2}
print(merged_dict)
```
