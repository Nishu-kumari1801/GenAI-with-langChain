### Introduction to tuples

### Explanation 
Tuples are ordered collections of items that are immutable . They are similar to lists , but their immutability makes them differebt

```python
## creating a tuple
empty_tuple=()
print(empty_tuple)
print(type(empty_tuple))
```

```python
lst = list()
print(type(lst))
tpl = tuple()
print(type(tpl))
```

```python
numbers = tuple([1,2,3,4,5,6])
numbers
```

```python
### Immutable nature of tuples 
### Tuples are immutable , meaning their elements cannot be changed onnce assigned

lst = [1,2,3,4,5]
print(lst)

lst[1]="nishu"
print(lst)
```

```python
## Tuple Methods
print(numbers.count(1))
print(numbers.index(3))
```

```python
## packing and unpacking tuple

##packing
packed_tuple = 1,"hello",3.14
print(packed_tuple)
```

```python
## unpacking a tuple
a,b,c=packed_tuple
print(a)
print(b)
print(c)
```

```python
## Unpacking with *
numbers = (1,2,3,4,5,6)
first,*middle,last = numbers
print(first)
print(middle)
print(last)
```

```python
## Nested Tuples
## Nested list
lst =[[1,2,3,4],[6,7,8,9,10],[1,"hello",3.14,"c"]]
lst[0][0:3] 
```

```python
lst =[[1,2,3,4],[6,7,8,9,10],(1,"hello",3.14,"c")]
lst[2][0:3] 
```

```python
nested_tuple =((1,2,3),("a","b","c"),(True,False))

## access the elements inside a tuple
print(nested_tuple[0])
print(nested_tuple[1][2])
```

```python
## iterating over nested tuples
for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item,end=" ")
    print()
```
