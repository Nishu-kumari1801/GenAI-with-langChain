#### loops

```python
## for loop
for i in range(5):
    print(i)
```

```python
for i in range(1,6):
    print(i)
```

```python
### range(start,end,step)
for i in range (1,10,2):
    print(i)
```

```python
for i in range (10,1,-3):
    print(i)
```

```python
str = "Nishu Jha"
for i in str:
    print(i)
```

```python
### while loop

''' The while loop continues to execute as long as the condition is True'''

cnt =0;
while(cnt<5):
    print(cnt)
    cnt+=1
```

```python
count = 0
while count%2 ==0 :
    print(count)
    count+=1
```

```python
## loop control statements

## break : exists the loop prematurely

for i in range (10):
    if(i==5):
        break 
    print(i)
```

```python
### continue
## the continue statements skips the current iteration and continues with the next

for i in range (10):
    if i%2 == 0:
        continue
    print(i)
```

```python
## pass
## the pass statemmentt is a null operation and it does nothing

for i in range(5):
    if i==3:
        pass
    print(i)
```

```python
#### nested loop :- loop inside a loop

for i in range(3):
    for j in range(2):
        print(f"i:{i} and j:{j}")
```

```python
## Examples - calculate the sum of first n natural numbers using a while and for loop

n=10
sum =0
while(n):
    sum+=n
    n-=1
print(sum)
```

```python
n=10
sum =0
for i in range(n+1):
    sum+=i
print(sum)
```

```python
## prime number between 1 and 100

for num in range(1,101):
    if num >1:
        for i in range(2,num):
            if num%i==0:
                break
        else:
            print(num)
```
