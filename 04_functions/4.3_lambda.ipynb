## Lambda Functions in python 

### Lambda functions are small anonymous functions  defined using  the lambda keyword. they can have any number of arguments but only one expression . they are commonly used for short operations or as arguments to highe-order functions

```python
# syntax
lambda arguments : expression
```

```python
def addition(a,b):
    return a+b
```

```python
addition(2,7)
```

```python
## lambda function
addition=lambda a,b : a+b

type(addition)
print(addition(5,6))
```

```python
def even (num):
    if num%2==0:
        return True
even(24)
```

```python
even1 = lambda num:num%2==0
even1(12)
even1(97)
```

```python
def addition(x,y,z):
    return x+y+z
addition(12,13,14)
```

```python
addition1 = lambda x,y,z:x+y+z
addition1(12,13,19)
```

```python
## map()
def square(number):
    return number**2
square(2)
```

```python
## map() - applies a function to all items in a list (funxtion,iterables)
numbers = [1,2,3,4,5,6]
list(map(lambda x:x**2,numbers))
```
