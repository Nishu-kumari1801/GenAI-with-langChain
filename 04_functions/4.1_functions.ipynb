## Introduction to function

### Defination
A function is a block of code that performs a specific task . Function help in organising code , reusing code and improving readibility .

```python
def function_name(parameters):
    """docstring"""
    #Function body
    return expression
```

```python
## why function
def evenOdd(num):
    """This function finds even or odd """
    if num%2==0:
        print("the number is even")
    else:
        print("the number is odd")

## call the function
evenOdd(24)
evenOdd(89)
```

```python
## function with multiple parameters
def add(a,b):
    return a+b

result=add(2,4)
print(result)
```

```python
## default parameters
def greet(name="Guest"):
    print(f"Hello {name} , welcome to the paradise")
greet("Nishu")
greet()
```

```python
### variablle length argumennts
## positional and keyword arguments

def print_numbers(*args):
    for numbers in args:
        print(numbers)

print_numbers(1,2,3,4,5,6,7,"nishu")
```

```python
### keywords arguments

def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")

print_details(name="nishu",age=34,student="true")
```

```python
## positional keyarguments

def print_details(*args,**kwargs):
    for val in args:
        print(f"Positional arguments : {val}")
    for key,value in kwargs.items():
        print(f"{key}:{value}")

print_details(1,2,3,4,5,"nishu",name="nishu",age=34,student="true")
```

```python
## return statements

def multiply(a,b):
    return a*b
multiply(8,9)
```

```python
## return multiple parameters from a variable
def multiply(a,b):
    c=a*b
    return c,a
multiply(7,5)
```
