## Tokenisation

```python
corpus = """ Hello Welcome , to nishu jha's platform .
Please go through the platform and learn something new . 
"""
```

```python
print(corpus)
```

```python
import nltk

nltk.download('punkt_tab')
```

```python
### tokenzation :- convert a sentence into paragraph (sentence->paragraph)
from nltk.tokenize import sent_tokenize
```

```python
documents = sent_tokenize(corpus)
```

```python
type(documents)
```

```python
for sentence in documents :
    print(sentence)
```

```python
## tokenization :- convert a paragraph into words (paragraph --> words , sentence-->words)
from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
for sentence in documents :
    print(word_tokenize(sentence))
```

```python
from nltk.tokenize import wordpunct_tokenize
```

```python
wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordTokenizer
```

```python
tokenizer = TreebankWordTokenizer()
```

```python
tokenizer.tokenize(corpus)
```
