## stemming 

Stemming is the process of reducing a word to its word stem that affixes to suffixes and prefixes or to the roots of words known as a lemma . Stemming is important in natural language understanding and natural language processing .

```python
## classification problem 
## comments of product is a positive review or negative review 
## reviews ----> [eating , eat, eaten]-->eat, [going, gone,goes]--->go

words =["eating","eats","eaten","writing","writes","programming","program","programs","history","finally","finalizes"]
```

## porterStemmer

```python
from nltk.stem import PorterStemmer
```

```python
stemming = PorterStemmer()
```

```python
for word in words:
    print(word+"---->"+stemming.stem(word))
```

```python
stemming.stem('congratulation')
```

```python
stemming.stem('sitting')
```

## RegexpStemmer class

NLTK has regexpStemmer class with the help of which we can easily implement Regular Expression Stemmer algorithms . It basically takes a single regular expression and removes  any prefix and suffix that matches the expression.

```python
from nltk.stem import RegexpStemmer
```

```python
reg_stemmer = RegexpStemmer('ing$|s$|e$|able$',min=4)
```

```python
reg_stemmer.stem('eating')
```

```python
reg_stemmer.stem('ingeating')
```

```python
reg_stemmer.stem('maintainable')
```

### Snowball Stemmer

Snowball Stemmer is a stemming algorithm in NLTK that reduces words to their root or base form by removing common prefixes and suffixes.

```python
from nltk.stem import SnowballStemmer 
```

```python
snowballstemmer = SnowballStemmer('english')
```

```python
for word in words:
    print(word+"---->"+snowballstemmer.stem(word))
```

```python
stemming.stem("fairly"),stemming.stem("sportingly")
```

```python
snowballstemmer.stem("fairly"),snowballstemmer.stem("sportingly")
```

```python
snowballstemmer.stem("going"),snowballstemmer.stem("goes")
```
